```python
import mlflow
import torch
```

```python
mlflow.set_tracking_uri("http://localhost:5001")

mlflow.set_experiment("FNN Classification")

run = mlflow.start_run(run_name="last_big_run_3")
```

```python
client = mlflow.tracking.MlflowClient()
```

```python
model = mlflow.pytorch.load_model("models:/FNN_classification_model/1")
```

```python
from sklearn.datasets import load_breast_cancer
import torch

data = load_breast_cancer()
X = data.data
y = data.target


from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4, random_state=42)


X_test  = torch.tensor(X_test, dtype=torch.float32)
y_test  = torch.tensor(y_test, dtype=torch.long)


from torch.utils.data import TensorDataset

test_dataset  = TensorDataset(X_test, y_test)



from torch.utils.data import DataLoader

test_loader  = DataLoader(test_dataset, batch_size=1, shuffle=False)
```

```python
X = next(iter(test_loader))[0]
y = next(iter(test_loader))[1]
```

```python
print(y)
```

```python



with torch.no_grad():

    predicted_correct, actual = [], []

    for i, (X, y) in enumerate(test_loader):

        output = model(X)

        prediction = torch.argmax(output, dim=1)

        print("predicted: ",prediction.item(), "    Actual Label: ",y.item())

        if prediction.item() == y.item():
            predicted_correct.append(y.item())

        j = i
```

```python
len(predicted_correct)/j
```
